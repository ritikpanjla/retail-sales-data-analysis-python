```python
r=df.groupby("Product")["Quantity"].sum()
re=r.sort_values(ascending=False)
print(re.head(1))
```

```python
df.groupby("Category")["Sales"].sum()
```

```python
categorical_analysis=df.groupby("Category").agg({"Sales":"sum", "Quantity":"sum"})
print(categorical_analysis)

categorical_analysis.sort_values(by="Sales", ascending=False)
```

```python
#City-Wise Sales
```

```python
city_sales= df.groupby("City")["Sales"].sum()
city_sales.sort_values(ascending=False)
```

```python
#Monthly Sales Analysis
```

```python
df["Date"]=pd.to_datetime(df["Date"])
```

```python
df["Month"]=df["Date"].dt.month_name()
```

```python
monthly_sales=df.groupby("Month")["Sales"].sum()
monthly_sales.sort_values(ascending=False)
```

```python
#Top 5 Best_Selling Products
```

```python
product_sales=df.groupby("Product")["Quantity"].sum()
top_5_products= product_sales.sort_values(ascending=False).head(5)
top_5_products
```

```python
#Data Visualization 
```

```python
import matplotlib.pyplot as plt
```

```python
top_5_products.plot(kind="bar")

plt.title("Top 5 Best Selling Products")
plt.xlabel("Product Name")
plt.ylabel("Quantity Sold")
plt.show()
```

```python
monthly_saless=df.groupby("Month")["Sales"].sum()
monthly_saless.plot(kind="line", marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month Name")
plt.ylabel("Sales")
plt.show()
```

```python
monthly_saless=df.groupby(df["Date"].dt.month)["Sales"].sum()
monthly_saless.plot(kind="line", marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month Name")
plt.ylabel("Total Sales")
plt.xticks([1, 2, 3, 4],[ "Jan", "Feb", "Mar", "Apr"])
plt.show()
```

```python
#City_wise Sales Pie Chart 
```

```python
city_sales.plot(kind="pie", autopct="%1.1f%%")
plt.title("City Wise Sales Distribution")
plt.ylabel("")
plt.show()
```
